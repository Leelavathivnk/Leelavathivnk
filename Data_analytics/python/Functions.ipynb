PYTHON -FUNCTION

```python
#CREATING A FUNCTION
def greet():
  print("hello")
greet()#calling function
```

```python
#FUNCTION WITH PARAMETER
def greet(name):
  print("hi",name)
greet("leela")
```

```python
#MULTIPLE PARAMETERS
def add(a,b):
  print(a+b)
add(10,5)
```

```python
#FUNTION WITH RETURN
def add(a,b):
  return a+b
result=add(20,30)
print(result)

def square(num):
  return num*num
result=square(5)
print(result)
```

```python
#FUNCTION WITH CALCULATION
def calculate_profit(sales,expenses):
  profit=sales-expenses
  return profit
result=calculate_profit(50000,32000)
print(result)
```

```python
#FUNTION+DA LOGIC
def check_marks(marks):
  if marks>=50:
    return "pass"
  else:
    return "fail"
result=check_marks(72)
print(result)
```
