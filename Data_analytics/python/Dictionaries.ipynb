DICTIONARIES

```python
#ACCESS A VALUE
student = {
    "name": "Leela",
    "age": 21,
    "course": "Data Analytics"
}
print(student["course"])
```

```python
# ACCESS USING KEY
employee = {
    "name": "Arun",
    "salary": 30000,
    "department": "IT"
}

print(employee["salary"])
```

```python
# ADD A NEW KEY
student = {
    "name": "Leela",
    "age": 21
}
student["city"] = "chennai"
print(student)
```

```python
# UPDATE A VALUE
employee = {
    "name": "Priya",
    "salary": 30000
}

employee["salary"] = 35000
print(employee)
```

```python
#FIND DICTIONARY LENGTH
product = {
    "name": "Laptop",
    "price": 50000,
    "brand": "HP",
    "stock": 10
}

print(len(product))
```

```python
#LOOP THROUGH KEYS
employee = {
    "name": "Priya",
    "age": 25,
    "salary": 35000
}

for key in employee:
  print(key)
```

```python
#Loop through values
employee = {
    "name": "Priya",
    "age": 25,
    "salary": 35000
}

for key in employee:
  print(employee[key])
```

```python
# Dictionary + condition
employees = {
    "Arun": 25000,
    "Priya": 35000,
    "Rahul": 28000,
    "Meena": 40000
}

for key in employees:
    if employees[key] > 30000:
       print(key)
```

```python
# Count using a dictionary
marks = {
    "Maths": 80,
    "Python": 75,
    "SQL": 90,
    "Excel": 65
}

for key in marks:
  if marks[key] >= 75:
    count=count+key
```
