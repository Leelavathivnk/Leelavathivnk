TUPLES -PRACTICE QUESTIONS

```python
# INDEXING
numbers=(10,20,30,40,50)
print(numbers[2])
```

```python
# NEGATIVE INDEXING
sales=(100,200,300,400,500)
print(sales[-1])
```

```python
#SLICING
marks=(75,82,91,68,88)
print(marks[1:4])
```

```python
# LEN()
subjects=("python","sql","excel","power bi")
print(len(subjects))
```

```python
#LOOP
numbers=(10,20,30,40,50)
for num in numbers:
  print(num)
```

```python
#LOOP + CONDITION
marks=(45,72,88,35,91)
for mark in marks:
  if mark>=50:
    print(mark)
```

```python
#COUNT()
numbers= (10,20,10,30,10,40)
print(numbers.count(10))
```

```python
#DA QUESTION
sales=(1200,800,2500,1500,3000)
for sale in sales:
  if sale>1500:
    print(sale)
```

```python
#CHALLENGE
sales=(1000,2000,1500,3000)
total=0
for sale in sales:
  total=total+sale
print(total)
```

```python
#MINI CHALLENGE
marks=(45,72,88,35,91,60)
count=0

for mark in marks:
  if mark >= 50:
    count=count+1

print(count)
```
