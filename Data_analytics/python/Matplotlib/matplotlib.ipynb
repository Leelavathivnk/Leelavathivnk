```python
#foramt string
plots=dict(
    c='black',
    markerfacecolor='white',
    linewidth=2,
    linestyle='--',
    marker='o',
    markeredgecolor='blue',
    markeredgewidth=4) 
plt.plot(x,y,**plots)

#title
plt.title('example_plot',
    fontsize=20,
    c='black',
   family='times new roman')

#line plot
plt.xlabel('year',c='b')
plt.ylabel('amount',c='b')

#subplot
plt.subplots_adjust(left=0.5,bottom=0.5)

plt.tick_params(axis='y',labelsize=10,labelcolor='g')
plt.tick_params(axis='x',labelsize=10,labelcolor='gold')

#grid
plt.grid(axis='x',c='yellow',ls='--')
plt.grid(axis='y',c='red',linestyle=':')
```

```python
import numpy as np
```

```python
x=np.array(['jan','feb','march','april','may'])
y=np.array([105,230,300,490,510])
```

```python
plt.plot(x,y)
plt.title('monthly profit',c='brown')
plt.xlabel('month')
plt.ylabel('profit')
```

```python
#pie chart
plt.pie(y,labels=x,
        autopct='%1.1f%%',#(auto-percentage)
        colors=['r','b','y','m','c'],
       shadow=True,
       explode=[1,0,0,0,0],
       startangle=45)
plt.title('data',c='olive')
```

```python
#scatter plot
x=np.array([12,10,22,4,7,92])
y=np.array([2007,3045,4456,4900,7000,3000])
plt.scatter(x,y,c='r',alpha=0.8,edgecolor='black')
```

```python
x=np.array([12,10,22,4,7,92])
y=np.array([2007,3045,4456,4900,7000,3000])
x1=np.array([120,103,221,40,72,92])
y1=np.array([207,345,456,490,700,300])
```

```python
plt.scatter(x,y,c='r',alpha=0.8,edgecolor='black')
plt.scatter(x1,y1,c='purple',alpha=0.5,edgecolor='b')
plt.title('sample data')
plt.legend(x)
```

```python
#histogram
plt.hist(x,bins=20,edgecolor='r')
```
