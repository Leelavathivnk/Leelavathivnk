OPERATORS

```python
#ARITHEMETIC OPERATORS
a=10
b=3
print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a//b)
print(a%b)
print(a**b)
```

```python
#COMPARISON OPERATOR
salary=50000
print(salary>30000)
print(salary<30000)
print(salary>=30000)
print(salary<=30000)
```

```python
#LOGICAL OPERATORS
age=25
is_student=True
print(age>20 or age <30)# both conditions true
print(age>40 or age<26)# only one condition is true
print(not is_student) # reverses true/false

```

```python
#ASSIGNMENT QPERTORS
score=20
score+=10
print(score)
```
