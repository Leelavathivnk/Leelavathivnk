NESTED LOOPS



```python
#Q- Write a nested loop to produce this output:
for i in range(1,3):        #outerloop
  for j in range(1,4):      #innerloop
    print(i,j)

#Outer loop = 2 times
#Inner loop = 3 times each
#Total = 2 × 3 = 6 executions
```

SOLVING EXERCISES

```python
#Exercise 1. Print first 10 natural numbers using
# Initialize the counter
count=1
# Iterate until count is greater than 10
while count<=10:
  print (count)
 # Increment the counter to move to the next number
  count+=1
```

```python
#Exercise 2. Display numbers from -10 to -1 using for loop
# Start at -10, stop before 0
for i in range(-10,0):
  print(i)
```

```python
#Exercise 3. Display a message “Done” after successful execution of for loop
#Python feature: the Loop-Else Clause.
for i in range(5):
    print(i)
 # This block executes only after the loop finishes naturally
else:
    print("Done")
```

```python
#Exercise 4. Calculate the sum of all numbers from 1 to N

# Take input from user and convert to integer
num=int(input("enter a number:"))
# Variable to store the sum
total=0
#Loop from 1 to n (num+1 is used because range is EXCLUSIVE as it is not given in question)
for i in range(1,num+1):
  total+=i
print("total=",total)

```

```python
#Exercise 5. Print multiplication table of a given number
num=int(input("Eter a number:"))
for i in range(1,11):
  print(num*i)
```

```python
#Exercise 6.Calculate the cube of all numbers from 1 to a given number
#To calculate a cube, use i * i * i or i ** 3
num=int(input("enter a number:"))
for i in range(1,num+1):
  print(i**3)
```

```python
#Exercise 7. Display numbers from a list using a loop
#1The number must be divisible by five.
#2If the number is greater than 150, skip it and move to the next.
#3If the number is greater than 500, stop the loop entirely.

numbers = [12, 75, 150, 180, 145, 525, 50]
for i in numbers:
  # Condition 3: Stop the loop if number > 500
  if i>500:
    break

  # Condition 2: Skip the number if > 150
  if i>150:
    continue

  # Condition 1: Print if divisible by 5
  if i%5==0:
    print(i)
```

```python
#Exercise 8. Count occurrences of a specific element in a list
list1 = [10, 20, 10, 30, 10, 40, 50]
target=10
count=0

for i in list1:
  if i==target:
    count+=1

print("target:",target,"count:",count)
```
