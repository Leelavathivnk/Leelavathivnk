PYTHON-LIST


```python
# LIST INDEXING
name=["leela","logesh","saima","venkat"]
print(name[0])
```

```python
#NEGATIVE INDEXING
sales=[100,200,300,400,500]
print(sales[-1])
```

```python
#LIST SLICING
number=[1,2,3,4,5,6,7,8,9]
print(number [1:4])
print(number [:-5])
print(number [2:])
print(number [0:4:2])# {start,stop,step}
print(number [::-1]) #reverse


```

```python
#CHANGING LIST ITEM (replace)
number=[10,20,40,40,50]
number[2]=30
print(number)
```

```python
#append()
name=["leela","logesh","saima","venkat"]
name.append("bujji")
print(name)
```

```python
#insert()
number=[10,20,40,50,60]
number.insert(2,30)
print(number)
```

```python
#remove
name=["leela","logesh","saima","venkat"]
name.remove("saima")
print(name)
```

```python
#len--> length
number=[10,20,40,50,60]
print(len(number))
```

```python
# POP()
names=["lee","log","leelo"]
names.pop(0)
print(names)
```

```python
#LIST + FOR LOOP
numbers=[10,20,30,40,50,60]

for num in numbers:
  print(num)
```

```python
#LIST + LOOP+ IF CONDITION
data=[101,202,303,404,505]

for id in data:
  if id >=200:
    print(id)
```

```python
# total values in list

sales=[100,1500,300,4000]

total=0

for sale in sales:
  total=total+sale
print(sale)
```

```python
# count values in list

sales=[200,205,400,350,600,700,405]
count=0

for sale in sales:
  if sale>=400:
    count=count+1
    print(count)
```

```python
#multiple list in list
num=[1,2,3,[5,6]]
print(num[3][1])
```
