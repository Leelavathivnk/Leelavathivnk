CONDITIONAL STATEMENTS

```python
#IF
age=int(input ("age:"))
if (age>=18):
   print("your are an adult")
```

```python
#IF-ELSE
salary=int(input("salary"))
if(salary>=30000):
  print("you are eligible for loan")
else:
  print("you are not eligible for loan")
```

```python
# PRACTICE Question [posituve or negative]
num=-5
if(num>=0):
  print("positive")
else:
  print("negative")
```

```python
#IF-ELIF-ELSE
marks=63
if(marks>=90):
  print("excellent")
elif(marks>=60 ):
  print("very good")
else:
  print("average -need improvement")
```

```python
# PRACTICE Question [sales classification]
sales=65000
if(sales>=100000):
  print("excellent")
elif(sales>=50000):
  print(" good")
else:
  print("Low")
```
