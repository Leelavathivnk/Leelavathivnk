PYTHON- STRINGS


```python
#STRING INDEXING
name="leela"
print(name[2])
```

```python
#NEGATIVE INDEXING
name="venkat"
print(name[-1])
```

```python
#STRING SLICING
name="logesh"
print(name[0:3])
print(name[2:])
print(name[:3])
print(name[1:5])
print(name[0:5])
print(name[1:4])
#negative index
print(name[-4:])
print(name[:-1])
print(name[::2])
```

```python
#STRING METHODS
#.upper()
name="saima"
print(name.upper())
```

```python
#.lower()
name="PAVI"
print(name.lower())
```

```python
#.strip()
name="  leela "
print(name.strip())
```

```python
#.replace()
name="leela"
print(name.replace("leela","feela"))
```

```python
#.split()
name="leelavathi is a good girl"
fun="games  enjoy"
print(name.split(" "))
print(fun.split(","))
```

```python
#.find()
name="love is good"
print(name.find("good"))
```

```python
#chain method
names="leela"
print(names[2].upper().split(","))
```

```python
# STRINGS ARE IMMUTABLE
city="chennai"
new_city=city.upper()
print(city)
print(new_city)
```
