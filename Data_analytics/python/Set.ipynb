```python
#create a set
numbers={10,20,20,30,10,40}
print(numbers)
```

```python
#Duplicate values
cities = {"Chennai", "Mumbai", "Chennai", "Delhi", "Mumbai"}

print(len(cities))
```

```python
#Adding a value -> takes exactly one arguement
cities = {"Chennai", "Mumbai"}

cities.add("Delhi")

print(cities)
```

```python
#Removing a value
cities = {"Chennai", "Mumbai", "Delhi"}

cities.remove("Mumbai")

print(cities)
```

```python
#Set + Loop
cities = {"Chennai", "Mumbai", "Delhi"}

for city in cities:
    print(city)
```

```python
#Set + Data Analytics
departments = ["IT", "HR", "IT", "Sales", "HR", "IT"]
unique = set(departments)
print(unique)
```

```python
#Set Operations
dept1 = {"IT", "HR", "Sales"}
dept2 = {"IT", "Finance", "HR"}

intersection=dept1 & dept2
print(intersection)

union=dept1 | dept2
print(union)

diff=dept1 - dept2
print(diff)

symmetric_diff=dept1 ^ dept2 #diff in either
print(symmetric_diff)
```

```python
#Data Analytics-style
sales = ["Chennai", "Mumbai", "Chennai", "Delhi", "Mumbai", "Chennai"]

unique_cities = set(sales)

print(len(unique_cities))
```

```python
#unique categories
numbers = [2, 1, 2, 3, 1] #list data
unique_numbers = set(numbers) #list->set data
print(unique_numbers)
```
